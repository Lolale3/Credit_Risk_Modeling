```python
import os
import pandas as pd

try:
    from IPython.display import display
except ImportError:  # running as a plain .py script
    display = print


loan_data = pd.read_csv('/content/loan_data_2007_2014.csv', low_memory=False)

loan_data.head()

loan_data.info()

na_col = loan_data.isnull().mean()
na_col[na_col>0.8]

loan_data.drop(columns=na_col[na_col>0.8].index, inplace = True)

# Quarantine outcome and cash-flow fields: used only to define the target and value decisions, never as features.
outcome_cols = ['loan_status', 'issue_d', 'term', 'grade', 'sub_grade', 'int_rate', 'funded_amnt', 'total_pymnt',
                'total_rec_prncp', 'total_rec_int', 'total_rec_late_fee', 'recoveries', 'collection_recovery_fee',
                'last_pymnt_d']
outcomes = loan_data[[c for c in outcome_cols if c in loan_data.columns]].copy()

not_known_at_approval = [
    # identifiers and free text
    'Unnamed: 0', 'id', 'member_id', 'emp_title', 'url', 'desc', 'title', 'zip_code', 'policy_code',
    # post-origination
    'funded_amnt', 'funded_amnt_inv', 'out_prncp', 'out_prncp_inv', 'total_pymnt', 'total_pymnt_inv',
    'total_rec_prncp', 'total_rec_int', 'total_rec_late_fee', 'recoveries', 'collection_recovery_fee',
    'last_pymnt_d', 'last_pymnt_amnt', 'next_pymnt_d', 'last_credit_pull_d', 'pymnt_plan',
    # lender pricing (int_rate is used for the profit cutoff, from `outcomes`)
    'int_rate', 'installment', 'sub_grade',
    # not populated for 2007-2011 vintages
    'tot_coll_amt', 'tot_cur_bal', 'total_rev_hi_lim', 'mths_since_last_major_derog',
    # geography and platform attributes
    'addr_state', 'initial_list_status', 'application_type',
]
loan_data.drop(columns=[c for c in not_known_at_approval if c in loan_data.columns], inplace=True)

loan_data.info()

loan_data['loan_status'].value_counts(normalize = True)

import numpy as np

good_statuses = ['Fully Paid', 'Does not meet the credit policy. Status:Fully Paid']
bad_statuses = ['Charged Off', 'Default', 'Late (31-120 days)', 'Does not meet the credit policy. Status:Charged Off']

loan_data['good_bad'] = np.where(loan_data['loan_status'].isin(good_statuses), 1,
                                 np.where(loan_data['loan_status'].isin(bad_statuses), 0, np.nan))

loan_data.drop(columns = ['loan_status'], inplace = True)

import matplotlib.pyplot as plt

snapshot = pd.to_datetime(outcomes['last_pymnt_d'], format='%b-%y').max()
issue_dt = pd.to_datetime(loan_data['issue_d'], format='%b-%y')
loan_data['issue_year'] = issue_dt.dt.year
loan_data['term_months'] = loan_data['term'].astype(str).str.extract(r'(\d+)')[0].astype(float)
months_on_book = (snapshot.year - issue_dt.dt.year) * 12 + (snapshot.month - issue_dt.dt.month)
loan_data['matured'] = months_on_book >= loan_data['term_months']

print('Data snapshot (latest payment date):', snapshot.strftime('%Y-%m'))

is36 = loan_data['term_months'] == 36
labelled = loan_data['good_bad'].notna()
vintages = pd.DataFrame({
    'loans': loan_data.groupby('issue_year').size(),
    '36_month': loan_data[is36].groupby('issue_year').size(),
    '36_month_final_outcome': loan_data[is36 & loan_data['matured'] & labelled].groupby('issue_year').size(),
    'bad_rate': 1 - loan_data[is36 & loan_data['matured'] & labelled].groupby('issue_year')['good_bad'].mean(),
}).fillna(0)
display(vintages)

fig, ax = plt.subplots(figsize=(9, 4))
ax.bar(vintages.index, vintages['36_month'], color='lightgrey', label='36-month loans')
ax.bar(vintages.index, vintages['36_month_final_outcome'], color='steelblue', label='with final outcome')
ax.set_ylabel('Loans')
ax2 = ax.twinx()
final = vintages[vintages['36_month_final_outcome'] > 0]
ax2.plot(final.index, final['bad_rate'], color='firebrick', marker='o', label='bad rate')
ax2.set_ylabel('Bad rate')
ax.set_title('Loans by issue year: only matured vintages have final outcomes')
ax.legend(loc='upper left')
ax2.legend(loc='upper center');

from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold, cross_val_score

TRAIN_END_YEAR, TEST_YEAR, RECENT_YEARS = 2011, 2012, [2013, 2014]

train_mask = is36 & loan_data['matured'] & labelled & (loan_data['issue_year'] <= TRAIN_END_YEAR)
test_mask = is36 & loan_data['matured'] & labelled & (loan_data['issue_year'] == TEST_YEAR)
recent_mask = is36 & loan_data['issue_year'].isin(RECENT_YEARS)

# term is 36 months for everyone in this population, so it is no longer a feature
split_only = ['good_bad', 'issue_year', 'term', 'term_months', 'matured']
X_train = loan_data.loc[train_mask].drop(columns=split_only)
X_test = loan_data.loc[test_mask].drop(columns=split_only)
X_recent = loan_data.loc[recent_mask].drop(columns=split_only)
y_train = loan_data.loc[train_mask, 'good_bad'].astype(int)
y_test = loan_data.loc[test_mask, 'good_bad'].astype(int)

print(f'Train  (issued {TRAIN_END_YEAR} and earlier): {len(X_train):,} loans, bad rate {1 - y_train.mean():.1%}')
print(f'Test   (issued {TEST_YEAR}, out-of-time):    {len(X_test):,} loans, bad rate {1 - y_test.mean():.1%}')
print(f'Recent (issued {RECENT_YEARS[0]}-{RECENT_YEARS[-1]}, no outcomes): {len(X_recent):,} loans')

def emp_length_converter(df, column):
    text = df[column].astype(str)
    df[column] = pd.to_numeric(text.str.extract(r'(\d+)')[0], errors='coerce')
    df.loc[text.str.contains('<', na=False), column] = 0
    # 'n/a' and blanks stay missing: unknown employment gets its own WoE bin

emp_length_converter(X_train, 'emp_length')

X_train['emp_length'].unique()

def credit_history_months(df):
    issue = pd.to_datetime(df['issue_d'], format = "%b-%y")
    earliest = pd.to_datetime(df['earliest_cr_line'], format = "%b-%y")
    # two-digit years: 'Jan-62' parses as 2062, so shift any credit line that starts after the loan back a century
    future = earliest > issue
    earliest.loc[future] = earliest.loc[future] - pd.DateOffset(years=100)
    df['mths_since_earliest_cr_line'] = (issue.dt.year - earliest.dt.year) * 12 + (issue.dt.month - earliest.dt.month)
    df.drop(columns = ['earliest_cr_line', 'issue_d'], inplace = True)

credit_history_months(X_train)

print(X_train['mths_since_earliest_cr_line'].describe())

def add_ratios(df):
    df['loan_to_income'] = (df['loan_amnt'] / df['annual_inc'].replace(0, np.nan)).replace([np.inf, -np.inf], np.nan)

add_ratios(X_train)

X_train_cat = X_train.select_dtypes(include = ['object', 'string']).copy()
X_train_num = X_train.select_dtypes(include = 'number').copy()

from scipy.stats import chi2_contingency

chi2_check = {}

for column in X_train_cat:
    chi, p, dof, ex = chi2_contingency(pd.crosstab(y_train, X_train_cat[column]))
    chi2_check.setdefault('Feature',[]).append(column)
    chi2_check.setdefault('p-value',[]).append(round(p, 10))

chi2_result = pd.DataFrame(data = chi2_check)
chi2_result.sort_values(by = ['p-value'], ascending = True, ignore_index = True, inplace = True)
chi2_result

from sklearn.feature_selection import f_classif

X_train_num.fillna(X_train_num.mean(), inplace = True)

# Calculate F Statistic and corresponding p values
F_statistic, p_values = f_classif(X_train_num, y_train)

ANOVA_F_table = pd.DataFrame(data = {'Numerical_Feature': X_train_num.columns.values, 'F-Score': F_statistic, 'p values': p_values.round(decimals=10)})
ANOVA_F_table.sort_values(by = ['F-Score'], ascending = False, ignore_index = True, inplace = True)
ANOVA_F_table


import matplotlib.pyplot as plt
import seaborn as sns

top_num_features = ANOVA_F_table.iloc[:20,0].to_list()

corrmat = X_train_num[top_num_features].corr()
plt.figure(figsize=(10,10))
sns.heatmap(corrmat);

drop_columns_list = chi2_result.iloc[4:, 0].to_list()

def col_to_drop(df, columns_list):
    df.drop(columns = [c for c in columns_list if c in df.columns], inplace = True)

col_to_drop(X_train, drop_columns_list)

for df in (X_test, X_recent):
    emp_length_converter(df, 'emp_length')
    credit_history_months(df)
    add_ratios(df)
    col_to_drop(df, drop_columns_list)

print(X_train.shape, X_test.shape, X_recent.shape)

X_train_prepr = X_train.copy()
y_train_prepr = y_train.copy()
X_test_prepr = X_test.copy()
y_test_prepr = y_test.copy()

def woe_discrete(df, cat_variabe_name, y_df):
    df = pd.concat([df[cat_variabe_name], y_df], axis = 1)
    df = pd.concat([df.groupby(df.columns.values[0], as_index = False, observed = True)[df.columns.values[1]].count(),
                    df.groupby(df.columns.values[0], as_index = False, observed = True)[df.columns.values[1]].mean()], axis = 1)
    df = df.iloc[:, [0, 1, 3]]
    df.columns = [df.columns.values[0], 'n_obs', 'prop_good']
    df['prop_n_obs'] = df['n_obs'] / df['n_obs'].sum()
    df['n_good'] = df['prop_good'] * df['n_obs']
    df['n_bad'] = (1 - df['prop_good']) * df['n_obs']
    df['prop_n_good'] = df['n_good'] / df['n_good'].sum()
    df['prop_n_bad'] = df['n_bad'] / df['n_bad'].sum()
    df['WoE'] = np.log(df['prop_n_good'] / df['prop_n_bad'])
    df = df.sort_values(['WoE'])
    df = df.reset_index(drop = True)
    df['diff_prop_good'] = df['prop_good'].diff().abs()
    df['diff_WoE'] = df['WoE'].diff().abs()
    df['IV'] = (df['prop_n_good'] - df['prop_n_bad']) * df['WoE']
    df['IV'] = df['IV'].sum()
    return df

sns.set()

def plot_by_woe(df_WoE, rotation_of_x_axis_labels = 0):
    x = np.array(df_WoE.iloc[:, 0].apply(str))
    y = df_WoE['WoE']
    plt.figure(figsize=(18, 6))
    plt.plot(x, y, marker = 'o', linestyle = '--', color = 'k')
    plt.xlabel(df_WoE.columns[0])
    plt.ylabel('Weight of Evidence')
    plt.title(str('Weight of Evidence by ' + df_WoE.columns[0]))
    plt.xticks(rotation = rotation_of_x_axis_labels)

df_temp = woe_discrete(X_train_prepr, 'grade', y_train_prepr)
df_temp

plot_by_woe(df_temp)

df_temp = woe_discrete(X_train_prepr, 'home_ownership', y_train_prepr)
df_temp

plot_by_woe(df_temp)

df_temp = woe_discrete(X_train_prepr, 'verification_status', y_train_prepr)
df_temp

plot_by_woe(df_temp)

df_temp = woe_discrete(X_train_prepr, 'purpose', y_train_prepr)
df_temp

plot_by_woe(df_temp, 90)

def woe_ordered_continuous(df, continuous_variabe_name, y_df):
    df = pd.concat([df[continuous_variabe_name], y_df], axis = 1)
    df = pd.concat([df.groupby(df.columns.values[0], as_index = False, observed = True)[df.columns.values[1]].count(),
                    df.groupby(df.columns.values[0], as_index = False, observed = True)[df.columns.values[1]].mean()], axis = 1)
    df = df.iloc[:, [0, 1, 3]]
    df.columns = [df.columns.values[0], 'n_obs', 'prop_good']
    df['prop_n_obs'] = df['n_obs'] / df['n_obs'].sum()
    df['n_good'] = df['prop_good'] * df['n_obs']
    df['n_bad'] = (1 - df['prop_good']) * df['n_obs']
    df['prop_n_good'] = df['n_good'] / df['n_good'].sum()
    df['prop_n_bad'] = df['n_bad'] / df['n_bad'].sum()
    df['WoE'] = np.log(df['prop_n_good'] / df['prop_n_bad'])
    #df = df.sort_values(['WoE'])
    #df = df.reset_index(drop = True)
    df['diff_prop_good'] = df['prop_good'].diff().abs()
    df['diff_WoE'] = df['WoE'].diff().abs()
    df['IV'] = (df['prop_n_good'] - df['prop_n_bad']) * df['WoE']
    df['IV'] = df['IV'].sum()
    return df

X_train_prepr['int_rate'] = outcomes.loc[X_train_prepr.index, 'int_rate']
X_train_prepr['int_rate_factor'] = pd.cut(X_train_prepr['int_rate'], 50)

df_temp = woe_ordered_continuous(X_train_prepr, 'int_rate_factor', y_train_prepr)
df_temp

plot_by_woe(df_temp, 90)

df_temp = woe_ordered_continuous(X_train_prepr, 'emp_length', y_train_prepr)
df_temp

plot_by_woe(df_temp)

X_train_prepr['annual_inc_factor'] = pd.cut(X_train_prepr['annual_inc'], 50)

df_temp = woe_ordered_continuous(X_train_prepr, 'annual_inc_factor', y_train_prepr)
df_temp

# Initial examination shows that there are too few individuals with large income (up to $75m!!) and too many with small income.
# Since 95.6% of observations have income < 151,858, we will have one category for more than 150K, and we are going to apply our approach to determine
# the categories of everyone with 150k or less.
X_train_prepr_temp = X_train_prepr[X_train_prepr['annual_inc'] <= 150000].copy()

X_train_prepr_temp['annual_inc_factor'] = pd.cut(X_train_prepr_temp['annual_inc'], 50)

df_temp = woe_ordered_continuous(X_train_prepr_temp, 'annual_inc_factor', y_train_prepr[X_train_prepr_temp.index])
df_temp

plot_by_woe(df_temp, 90)

X_train_prepr['dti_factor'] = pd.cut(X_train_prepr['dti'], 50)

df_temp = woe_ordered_continuous(X_train_prepr, 'dti_factor', y_train_prepr)
df_temp

plot_by_woe(df_temp, 90)

df_temp = woe_ordered_continuous(X_train_prepr, 'inq_last_6mths', y_train_prepr)
df_temp

plot_by_woe(df_temp)

# revol_util is stored in percent (e.g. 83.7), not as a fraction. v1 filtered to <= 1, which kept only about 1% of loans.
# Values above 100% are rare, so we filter them out for this plot.
X_train_prepr_temp = X_train_prepr[X_train_prepr['revol_util'] <= 100].copy()

X_train_prepr_temp['revol_util_factor'] = pd.cut(X_train_prepr_temp['revol_util'], 10)

df_temp = woe_ordered_continuous(X_train_prepr_temp, 'revol_util_factor', y_train_prepr[X_train_prepr_temp.index])
df_temp

plot_by_woe(df_temp, 90)

# Initial examination shows that there are some obs >50 accounts which should be very rare, so we will filter them out first
X_train_prepr_temp = X_train_prepr[X_train_prepr['total_acc'] <= 50].copy()

X_train_prepr_temp['total_acc_factor'] = pd.cut(X_train_prepr_temp['total_acc'], 20)

df_temp = woe_ordered_continuous(X_train_prepr_temp, 'total_acc_factor', y_train_prepr[X_train_prepr_temp.index])
df_temp

plot_by_woe(df_temp, 90)

X_train_prepr['mths_since_earliest_cr_line_factor'] = pd.cut(X_train_prepr['mths_since_earliest_cr_line'], 25)

df_temp = woe_ordered_continuous(X_train_prepr, 'mths_since_earliest_cr_line_factor', y_train_prepr)
df_temp

plot_by_woe(df_temp, 90)

from sklearn.pipeline import Pipeline
from sklearn.base import BaseEstimator, TransformerMixin

NUMERIC_FEATURES = ['loan_amnt', 'annual_inc', 'loan_to_income', 'dti', 'emp_length', 'mths_since_earliest_cr_line',
                    'delinq_2yrs', 'inq_last_6mths', 'mths_since_last_delinq', 'mths_since_last_record', 'open_acc',
                    'pub_rec', 'revol_bal', 'revol_util', 'total_acc', 'collections_12_mths_ex_med', 'acc_now_delinq']
CATEGORICAL_FEATURES = ['grade', 'home_ownership', 'verification_status', 'purpose']

# v1 groupings, decided from the WoE plots
CATEGORY_GROUPS = {
    'home_ownership': {'OTHER': 'OTHER_NONE_RENT', 'NONE': 'OTHER_NONE_RENT', 'RENT': 'OTHER_NONE_RENT'},
    'purpose': {
        'major_purchase': 'major_purch__car__home_impr', 'car': 'major_purch__car__home_impr',
        'home_improvement': 'major_purch__car__home_impr',
        'educational': 'educ__ren_en__sm_b__mov', 'renewable_energy': 'educ__ren_en__sm_b__mov',
        'small_business': 'educ__ren_en__sm_b__mov', 'moving': 'educ__ren_en__sm_b__mov',
        'vacation': 'vacation__house__wedding__med__oth', 'house': 'vacation__house__wedding__med__oth',
        'wedding': 'vacation__house__wedding__med__oth', 'medical': 'vacation__house__wedding__med__oth',
        'other': 'vacation__house__wedding__med__oth',
    },
}


def _fmt(v):
    return f'{v:,.4g}'


class WoE_Binning(BaseEstimator, TransformerMixin):
    # Fits WoE bins on training data, then one-hot encodes each loan's bin (reference bins dropped).

    def __init__(self, numeric_features=None, categorical_features=None, max_bins=8,
                 min_bin_share=0.05, min_cat_share=0.02, iv_min=0.02, corr_max=0.75):
        self.numeric_features = numeric_features
        self.categorical_features = categorical_features
        self.max_bins = max_bins
        self.min_bin_share = min_bin_share
        self.min_cat_share = min_cat_share
        self.iv_min = iv_min
        self.corr_max = corr_max

    # ---------- fitting ----------
    def fit(self, X, y):
        bad = 1 - np.asarray(y, dtype=float)          # y is good_bad (1 = good)
        self.bins_ = {}
        for f in (self.numeric_features or []):
            if f in X.columns:
                self.bins_[f] = self._fit_numeric(X[f], bad)
        for f in (self.categorical_features or []):
            if f in X.columns:
                self.bins_[f] = self._fit_categorical(f, X[f], bad)
        self.iv_ = {f: b['iv'] for f, b in self.bins_.items()}

        self.dropped_low_iv_ = sorted(f for f, v in self.iv_.items() if v < self.iv_min)
        candidates = [f for f in self.bins_ if self.iv_[f] >= self.iv_min]
        woe_values = pd.DataFrame({f: self._woe_values(X[f], f) for f in candidates}, index=X.index)
        corr = woe_values.corr().abs().fillna(0)
        kept, self.dropped_corr_ = [], []
        for f in sorted(candidates, key=lambda f: -self.iv_[f]):
            partners = [k for k in kept if corr.loc[f, k] > self.corr_max]
            if partners:
                self.dropped_corr_.append((f, partners[0], round(float(corr.loc[f, partners[0]]), 2)))
            else:
                kept.append(f)
        self.selected_ = [f for f in candidates if f in kept]

        self.ref_categories_, self.columns_ = [], []
        for f in self.selected_:
            b = self.bins_[f]
            ref = b['labels'][int(np.argmin(b['woe']))]
            self.ref_categories_.append(f'{f}:{ref}')
            self.columns_ += [f'{f}:{lab}' for lab in b['labels'] if lab != ref]

        tables = []
        for f, b in self.bins_.items():
            t = b['table'].copy()
            t.insert(0, 'feature', f)
            t['IV'] = b['iv']
            t['selected'] = f in self.selected_
            tables.append(t)
        self.woe_table_ = pd.concat(tables, ignore_index=True)
        return self

    def _fit_numeric(self, x, bad):
        xv = pd.to_numeric(x, errors='coerce').to_numpy(dtype=float)
        nn = ~np.isnan(xv)
        total = len(xv)
        edges = []
        if nn.sum() > 0:
            qs = np.quantile(xv[nn], np.linspace(0, 1, self.max_bins + 1)[1:-1])
            edges = [e for e in np.unique(qs) if e < np.nanmax(xv)]

        def counts(e):
            idx = np.searchsorted(e, xv[nn], side='left')
            k = len(e) + 1
            return (np.bincount(idx, minlength=k).astype(float),
                    np.bincount(idx, weights=bad[nn], minlength=k))

        # merge bins that are too small into the neighbour with the closer bad rate
        while len(edges) > 0:
            n, b = counts(edges)
            small = np.where(n / total < self.min_bin_share)[0]
            if len(small) == 0:
                break
            i = small[np.argmin(n[small])]
            k = len(n)
            if i == 0:
                del edges[0]
            elif i == k - 1:
                del edges[k - 2]
            elif n[i] == 0:
                del edges[i - 1]
            else:
                br = np.divide(b, n, out=np.full_like(b, np.nan), where=n > 0)
                left, right = abs(br[i] - br[i - 1]), abs(br[i] - br[i + 1])
                if np.isnan(right) or (not np.isnan(left) and left <= right):
                    del edges[i - 1]
                else:
                    del edges[i]

        # merge neighbours until the bad rate is monotonic
        if len(edges) > 0:
            n, b = counts(edges)
            br = b / n
            if len(n) > 2:
                r = np.corrcoef(np.arange(len(n)), br)[0, 1]
                direction = np.sign(r) if np.isfinite(r) and r != 0 else 1.0
            else:
                direction = np.sign(br[-1] - br[0]) or 1.0
            while len(edges) > 0:
                n, b = counts(edges)
                violations = np.where(np.diff(b / n) * direction < 0)[0]
                if len(violations) == 0:
                    break
                del edges[violations[0]]

        edges = np.array(edges, dtype=float)
        n, b = counts(list(edges)) if nn.sum() > 0 else (np.array([0.0]), np.array([0.0]))
        if len(edges) == 0:
            labels = ['all']
        else:
            labels = ([f'<={_fmt(edges[0])}']
                      + [f'{_fmt(lo)}-{_fmt(hi)}' for lo, hi in zip(edges[:-1], edges[1:])]
                      + [f'>{_fmt(edges[-1])}'])
        n, b = list(n), list(b)
        if (~nn).any():
            labels.append('missing')
            n.append(float((~nn).sum()))
            b.append(float(bad[~nn].sum()))
        out = self._woe(labels, np.array(n), np.array(b), bad)
        out.update(kind='numeric', edges=edges)
        return out

    def _clean_cat(self, f, x):
        s = x.astype('string').str.strip().fillna('missing').replace('', 'missing').astype(object)
        return s.replace(CATEGORY_GROUPS.get(f, {}))

    def _fit_categorical(self, f, x, bad):
        s = self._clean_cat(f, x)
        share = s.value_counts(normalize=True)
        min_share = 0.0 if f == 'grade' else self.min_cat_share   # keep every grade, including G
        keep = set(share[share >= min_share].index)
        s = s.where(s.isin(keep), 'RARE')
        labels = sorted(s.unique())
        n = np.array([float((s == lab).sum()) for lab in labels])
        b = np.array([float(bad[(s == lab).to_numpy()].sum()) for lab in labels])
        out = self._woe(labels, n, b, bad)
        out.update(kind='categorical', keep=keep)
        return out

    @staticmethod
    def _woe(labels, n, b, bad):
        g = n - b
        G, B, k = float((bad == 0).sum()), float((bad == 1).sum()), len(labels)
        pg = (g + 0.5) / (G + 0.5 * k)      # +0.5 smoothing keeps WoE finite for empty cells
        pb = (b + 0.5) / (B + 0.5 * k)
        woe = np.log(pg / pb)
        table = pd.DataFrame({'bin': labels, 'n_obs': n.astype(int), 'prop_n_obs': n / n.sum(),
                              'bad_rate': np.divide(b, n, out=np.full_like(b, np.nan), where=n > 0),
                              'WoE': woe})
        return {'labels': labels, 'woe': woe, 'iv': float(np.sum((pg - pb) * woe)), 'table': table}

    # ---------- applying ----------
    def bin_labels(self, X, f):
        b = self.bins_[f]
        if b['kind'] == 'numeric':
            xv = pd.to_numeric(X[f], errors='coerce').to_numpy(dtype=float)
            nn = ~np.isnan(xv)
            n_value_bins = len(b['edges']) + 1
            labs = np.empty(len(xv), dtype=object)
            labs[nn] = np.array(b['labels'][:n_value_bins], dtype=object)[np.searchsorted(b['edges'], xv[nn], side='left')]
            labs[~nn] = 'missing'
            return labs
        s = self._clean_cat(f, X[f])
        return s.where(s.isin(b['keep']), 'RARE').to_numpy(dtype=object)

    def _woe_values(self, x, f):
        lookup = dict(zip(self.bins_[f]['labels'], self.bins_[f]['woe']))
        return np.array([lookup.get(l, 0.0) for l in self.bin_labels(x.to_frame(f), f)])

    def transform(self, X):
        cols = {}
        refs = set(self.ref_categories_)
        for f in self.selected_:
            labs = self.bin_labels(X, f)
            for lab in self.bins_[f]['labels']:
                name = f'{f}:{lab}'
                if name not in refs:
                    cols[name] = (labs == lab).astype(int)
        # a bin never seen in training gets all zeros, i.e. it is scored as the reference (riskiest) bin
        return pd.DataFrame(cols, index=X.index)[self.columns_]

print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
print(X_recent.shape)

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, roc_auc_score, confusion_matrix, precision_recall_curve, auc

reg = LogisticRegression(max_iter=1000)   # v2: no class_weight, so predicted probabilities stay calibrated
woe_transform = WoE_Binning(numeric_features=NUMERIC_FEATURES, categorical_features=CATEGORICAL_FEATURES)
pipeline = Pipeline(steps=[('woe', woe_transform), ('model', reg)])

# define cross-validation criteria. RepeatedStratifiedKFold automatially takes care of the class imbalance while splitting
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=1)

scores = cross_val_score(pipeline, X_train, y_train, scoring = 'roc_auc', cv = cv)
AUROC = np.mean(scores)
GINI = AUROC * 2 - 1

print('Mean AUROC (in-time, cross-validated): %.4f' % (AUROC))
print('Gini: %.4f' % (GINI))

pipeline.fit(X_train, y_train)

X_train_woe_transformed = pipeline['woe'].transform(X_train)   # transform only: the bins were fitted by pipeline.fit

feature_name = X_train_woe_transformed.columns.values

summary_table = pd.DataFrame(columns = ['Feature name'], data = feature_name)

summary_table['Coefficients'] = np.transpose(pipeline['model'].coef_)

summary_table.index = summary_table.index + 1
summary_table.loc[0] = ['Intercept', pipeline['model'].intercept_[0]]

summary_table.sort_index(inplace = True)
summary_table

woe_fitted = pipeline['woe']
iv = pd.Series(woe_fitted.iv_).sort_values()
colors = ['steelblue' if f in woe_fitted.selected_ else 'lightgrey' for f in iv.index]
plt.figure(figsize=(9, 6))
plt.barh(iv.index, iv.values, color=colors)
plt.axvline(woe_fitted.iv_min, color='k', linestyle='--', linewidth=1)
plt.xlabel('Information Value')
plt.title('Information Value by feature (grey = dropped)')
plt.tight_layout()

print('Selected:', woe_fitted.selected_)
print('Dropped, IV below', woe_fitted.iv_min, ':', woe_fitted.dropped_low_iv_)
print('Dropped, correlated with a stronger feature:', woe_fitted.dropped_corr_)
print('IV above 0.5 (leakage check):', [f for f, v in woe_fitted.iv_.items() if v > 0.5] or 'none')
woe_fitted.woe_table_[woe_fitted.woe_table_['selected']].round(4)

y_pred_test = pipeline.predict(X_test)

y_pred_test_proba = pipeline.predict_proba(X_test)

y_pred_test_proba = y_pred_test_proba[:][: , 1]

# we will now create a new DF with actual classes and the predicted probabilities
# create a temp y_test DF to reset its index to allow proper concaternation with y_hat_test_proba
y_test_temp = y_test.copy()
y_test_temp.reset_index(drop = True, inplace = True)
y_test_proba = pd.concat([y_test_temp, pd.DataFrame(y_pred_test_proba)], axis = 1)

# check the shape to make sure the number of rows is same as that in y_test
y_test_proba.shape

y_test_proba.columns = ['y_test_class_actual', 'y_hat_test_proba']

y_test_proba.index = X_test.index
y_test_proba['PD'] = 1 - y_test_proba['y_hat_test_proba']   # probability of default
y_test_proba.head()

# assign a threshold value to differentiate good with bad
tr = 0.5

y_test_proba['y_test_class_predicted'] = np.where(y_test_proba['y_hat_test_proba'] > tr, 1, 0)

confusion_matrix(y_test_proba['y_test_class_actual'], y_test_proba['y_test_class_predicted'], normalize = 'all')

fpr, tpr, thresholds_roc = roc_curve(y_test_proba['y_test_class_actual'], y_test_proba['y_hat_test_proba'])
plt.plot(fpr, tpr)

plt.plot(fpr, fpr, linestyle = '--', color = 'k')
plt.xlabel('False positive rate')
plt.ylabel('True positive rate')
plt.title('ROC curve (out-of-time)');

AUROC = roc_auc_score(y_test_proba['y_test_class_actual'], y_test_proba['y_hat_test_proba'])
AUROC

Gini = AUROC * 2 - 1
Gini

y_te = y_test_proba['y_test_class_actual'].to_numpy()
p_good_te = y_test_proba['y_hat_test_proba'].to_numpy()

KS = np.max(tpr - fpr)

rng = np.random.default_rng(7)
boot = []
for _ in range(300):
    i = rng.integers(0, len(y_te), len(y_te))
    boot.append(2 * roc_auc_score(y_te[i], p_good_te[i]) - 1)
gini_ci = np.percentile(boot, [2.5, 97.5])

p_good_tr = pipeline.predict_proba(X_train)[:, 1]
gini_train = 2 * roc_auc_score(y_train, p_good_tr) - 1

sub = outcomes.loc[X_test.index, 'sub_grade'].astype(str).str.strip()
sub_rank = sub.str[0].map({c: i for i, c in enumerate('ABCDEFG')}) * 5 + pd.to_numeric(sub.str[1:], errors='coerce')
ok = sub_rank.notna().to_numpy()
gini_champion = 2 * roc_auc_score(y_te[ok], -sub_rank[ok]) - 1   # better sub-grade = more likely good

print(f'Out-of-time Gini {Gini:.3f} (95% CI {gini_ci[0]:.3f} to {gini_ci[1]:.3f}), KS {KS:.3f}')
print(f'Train Gini {gini_train:.3f} (a large gap would signal overfitting)')
print(f'Champion, LendingClub sub-grade, out-of-time Gini: {gini_champion:.3f}')

fpr_c, tpr_c, _ = roc_curve(y_te[ok], -sub_rank[ok])
plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f'Scorecard (Gini {Gini:.2f})')
plt.plot(fpr_c, tpr_c, label=f'LendingClub sub-grade (Gini {gini_champion:.2f})')
plt.plot([0, 1], [0, 1], linestyle='--', color='k')
plt.xlabel('False positive rate')
plt.ylabel('True positive rate')
plt.title('Out-of-time ROC: scorecard vs champion')
plt.legend();

no_skill = y_test.mean()   # share of good loans in the test set
plt.plot([0, 1], [no_skill, no_skill], linestyle='--', label='No Skill')


precision, recall, thresholds_pr = precision_recall_curve(y_test_proba['y_test_class_actual'], y_test_proba['y_hat_test_proba'])

plt.plot(recall, precision, marker='.', label='Logistic')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.legend()
plt.title('PR curve');

auc_pr = auc(recall, precision)
auc_pr

calib = pd.DataFrame({'PD': y_test_proba['PD'], 'bad': 1 - y_test_proba['y_test_class_actual']})
calib['decile'] = pd.qcut(calib['PD'], 10, labels=False, duplicates='drop') + 1
calib_table = calib.groupby('decile').agg(loans=('bad', 'size'), mean_predicted_PD=('PD', 'mean'),
                                          observed_bad_rate=('bad', 'mean'))
display(calib_table.round(4))
print(f"Overall: mean predicted PD {calib['PD'].mean():.1%} vs observed bad rate {calib['bad'].mean():.1%}")

lim = calib_table[['mean_predicted_PD', 'observed_bad_rate']].max().max() * 1.05
plt.figure(figsize=(5, 5))
plt.plot(calib_table['mean_predicted_PD'], calib_table['observed_bad_rate'], marker='o')
plt.plot([0, lim], [0, lim], linestyle='--', color='k')
plt.xlabel('Mean predicted PD')
plt.ylabel('Observed bad rate')
plt.title('Calibration by decile (out-of-time)');

ref_categories = pipeline['woe'].ref_categories_   # fitted: the riskiest bin of each selected feature

df_ref_categories = pd.DataFrame(ref_categories, columns = ['Feature name'])
df_ref_categories['Coefficients'] = 0

df_scorecard = pd.concat([summary_table, df_ref_categories])
df_scorecard.reset_index(inplace = True)

df_scorecard['Original feature name'] = df_scorecard['Feature name'].str.split(':').str[0]

# Define the min and max threshholds for our scorecard
min_score = 300
max_score = 850

min_sum_coef = df_scorecard.groupby('Original feature name')['Coefficients'].min().sum()
max_sum_coef = df_scorecard.groupby('Original feature name')['Coefficients'].max().sum()

# create a new columns that has the imputed calculated Score based on the multiplication of the coefficient by the ratio of the differences between
# maximum & minimum score and maximum & minimum sum of cefficients.
df_scorecard['Score - Calculation'] = df_scorecard['Coefficients'] * (max_score - min_score) / (max_sum_coef - min_sum_coef)

# update the calculated score of the Intercept (i.e. the default score for each loan)
df_scorecard.loc[0, 'Score - Calculation'] = ((df_scorecard.loc[0,'Coefficients'] - min_sum_coef) / (max_sum_coef - min_sum_coef)) * (max_score - min_score) + min_score

df_scorecard['Score - Preliminary'] = df_scorecard['Score - Calculation'].round()

df_scorecard

# check the min and max possible scores of our scorecard
min_sum_score_prel = df_scorecard.groupby('Original feature name')['Score - Preliminary'].min().sum()
max_sum_score_prel = df_scorecard.groupby('Original feature name')['Score - Preliminary'].max().sum()

print(min_sum_score_prel)
print(max_sum_score_prel)

# v1 fixed the rounding overshoot by hand (intercept set to 598). Here the same fix is computed:
# shift the intercept so the lowest possible score is exactly min_score.
df_scorecard['Score - Final'] = df_scorecard['Score - Preliminary']
overshoot = df_scorecard.groupby('Original feature name')['Score - Final'].min().sum() - min_score
df_scorecard.loc[0, 'Score - Final'] = df_scorecard.loc[0, 'Score - Final'] - overshoot
df_scorecard

# Recheck min and max possible scores
print(df_scorecard.groupby('Original feature name')['Score - Final'].min().sum())
print(df_scorecard.groupby('Original feature name')['Score - Final'].max().sum())

X_test_woe_transformed = pipeline['woe'].transform(X_test)   # v1 called fit_transform here; with fitted bins that would refit on test data
X_test_woe_transformed.insert(0, 'Intercept', 1)
X_test_woe_transformed.head()

scorecard_scores = df_scorecard['Score - Final']

# Align the test columns to the scorecard rows by name. Reference categories have no dummy column and score 0.
X_test_woe_transformed = X_test_woe_transformed.reindex(columns = df_scorecard['Feature name'], fill_value = 0)

scorecard_scores = scorecard_scores.values.reshape(len(scorecard_scores), 1)
print(X_test_woe_transformed.shape)
print(scorecard_scores.shape)

# matrix dot multiplication of test set with scorecard scores
y_scores = X_test_woe_transformed.dot(scorecard_scores)
y_scores.head()

def score_loans(X):
    # Scorecard points for any set of loans: the same steps as above, wrapped for reuse
    d = pipeline['woe'].transform(X)
    d.insert(0, 'Intercept', 1)
    d = d.reindex(columns = df_scorecard['Feature name'], fill_value = 0)
    return pd.Series(d.to_numpy() @ df_scorecard['Score - Final'].to_numpy(), index = X.index)

scores_train, scores_test, scores_recent = score_loans(X_train), score_loans(X_test), score_loans(X_recent)
scores_test.describe()

# Calculate Youden's J-Statistic to identify the best threshhold.
# v2: computed on the TRAINING predictions, so the test set plays no part in choosing it.
fpr_tr, tpr_tr, thresholds_roc_tr = roc_curve(y_train, p_good_tr)
J = tpr_tr - fpr_tr

# locate the index of the largest J
ix = np.argmax(J)
best_thresh = thresholds_roc_tr[ix]

print('Best Threshold: %f' % (best_thresh))

# update the threshold value
tr = best_thresh

# crate a new column for the predicted class based on predicted probabilities and threshold
y_test_proba['y_test_class_predicted'] = np.where(y_test_proba['y_hat_test_proba'] > tr, 1, 0)

confusion_matrix(y_test_proba['y_test_class_actual'], y_test_proba['y_test_class_predicted'], normalize = 'all')

# create a new DF comprising of the thresholds from the out-of-time ROC output (dropping the infinite first threshold)
df_cutoffs = pd.DataFrame(thresholds_roc, columns = ['thresholds'])
df_cutoffs = df_cutoffs[(df_cutoffs['thresholds'] > 0) & (df_cutoffs['thresholds'] < 1)].reset_index(drop = True)

# calcue Score corresponding to each threshold
df_cutoffs['Score'] = ((np.log(df_cutoffs['thresholds'] / (1 - df_cutoffs['thresholds'])) - min_sum_coef) *
                       ((max_score - min_score) / (max_sum_coef - min_sum_coef)) + min_score).round()
df_cutoffs.head()

def n_approved(p):
    return np.where(y_test_proba['y_hat_test_proba'] >= p, 1, 0).sum()

df_cutoffs['N Approved'] = df_cutoffs['thresholds'].apply(n_approved)
df_cutoffs['N Rejected'] = y_test_proba['y_hat_test_proba'].shape[0] - df_cutoffs['N Approved']

df_cutoffs['Approval Rate'] = df_cutoffs['N Approved'] / y_test_proba['y_hat_test_proba'].shape[0]
df_cutoffs['Rejection Rate'] = 1 - df_cutoffs['Approval Rate']

df_cutoffs.head()

# the approval and rejection rates at the Youden threshold
df_cutoffs.iloc[(df_cutoffs['thresholds'] - best_thresh).abs().argsort()[:1]]

# compare the above rates with the case of the default 0.5 threshold
df_cutoffs.iloc[(df_cutoffs['thresholds'] - 0.5).abs().argsort()[:1]]

def realized_profit(index):
    # Cash collected minus cash lent. Evaluation only: these fields are never features.
    o = outcomes.loc[index]
    cash_in = (o['total_rec_prncp'].fillna(0) + o['total_rec_int'].fillna(0) + o['total_rec_late_fee'].fillna(0)
               + o['recoveries'].fillna(0) - o['collection_recovery_fee'].fillna(0))
    return cash_in - o['funded_amnt']

lent_train = outcomes.loc[X_train.index, 'funded_amnt']
profit_train = realized_profit(X_train.index)
good_tr = (y_train == 1).to_numpy()

g = profit_train[good_tr].sum() / lent_train[good_tr].sum()
l = -profit_train[~good_tr].sum() / lent_train[~good_tr].sum()
pd_star = g / (g + l)

bad_tr = outcomes.loc[X_train.index[~good_tr]]
ead = (bad_tr['funded_amnt'] - bad_tr['total_rec_prncp']).sum()
lgd = 1 - (bad_tr['recoveries'].fillna(0) - bad_tr['collection_recovery_fee'].fillna(0)).sum() / ead

print(f'Return on a repaid loan, g:     {g:.1%} of amount lent')
print(f'Net loss on a defaulted loan, l: {l:.1%} of amount lent (EAD {ead / lent_train[~good_tr].sum():.0%} of amount, LGD {lgd:.0%})')
print(f'Flat profit rule: approve if PD < g/(g+l) = {pd_star:.1%}')

rate_train = outcomes.loc[X_train.index, 'int_rate'].to_numpy(dtype=float)
return_train = (profit_train / lent_train).to_numpy()
okr = np.isfinite(rate_train) & np.isfinite(return_train)
g_coef = np.polyfit(rate_train[okr & good_tr], return_train[okr & good_tr], 1)
l_coef = np.polyfit(rate_train[okr & ~good_tr], -return_train[okr & ~good_tr], 1)
rate_fill = np.nanmedian(rate_train)

def price_aware_cutoff(rate):
    rate = np.where(np.isfinite(rate), rate, rate_fill)
    g_r = np.clip(np.polyval(g_coef, rate), 1e-3, None)
    l_r = np.clip(np.polyval(l_coef, rate), 1e-2, None)
    return g_r / (g_r + l_r)

rates = np.linspace(np.nanpercentile(rate_train, 2), np.nanpercentile(rate_train, 98), 50)
plt.figure(figsize=(7, 4))
plt.plot(rates, price_aware_cutoff(rates), label='Price-aware cutoff')
plt.axhline(pd_star, color='grey', linestyle='--', label='Flat cutoff')
plt.xlabel('Interest rate (%)')
plt.ylabel('Maximum PD to approve')
plt.title('The cutoff rises with price')
plt.legend();

PD_test = y_test_proba['PD'].to_numpy()
bad_te = 1 - y_te
profit_test = realized_profit(X_test.index).to_numpy()
rate_test = outcomes.loc[X_test.index, 'int_rate'].to_numpy(dtype=float)
profit_all = profit_test.sum()

order = np.argsort(PD_test)
cum_profit = np.cumsum(profit_test[order])
k_best = int(np.argmax(cum_profit))
hindsight_thr = (PD_test[order][k_best] + PD_test[order][min(k_best + 1, len(order) - 1)]) / 2

policies = {
    'Approve all (LendingClub book)': np.ones(len(PD_test), bool),
    'PD < 0.5': PD_test < 0.5,
    "Youden's J (fit on train)": p_good_te >= best_thresh,
    'Flat profit rule (fit on train)': PD_test < pd_star,
    'Price-aware profit rule (fit on train)': PD_test < price_aware_cutoff(rate_test),
    'Best in hindsight (reference only)': PD_test < hindsight_thr,
}

rows = []
for name, approve in policies.items():
    p = profit_test[approve].sum()
    rows.append({
        'policy': name,
        'approval rate': approve.mean(),
        'bad rate of approved': bad_te[approve].mean(),
        'good loans declined (% of goods)': ((bad_te == 0) & ~approve).sum() / (bad_te == 0).sum(),
        'bad loans declined (% of bads)': ((bad_te == 1) & ~approve).sum() / (bad_te == 1).sum(),
        'realized profit ($M)': p / 1e6,
        'vs approve all': (p - profit_all) / abs(profit_all),
    })
policy_table = pd.DataFrame(rows).set_index('policy')

def uplift_ci(approve, n_boot=300):
    # Uplift vs approve-all equals minus the realized profit of the loans declined.
    declined = ~approve
    boots = []
    for _ in range(n_boot):
        i = rng.integers(0, len(profit_test), len(profit_test))
        boots.append(-profit_test[i][declined[i]].sum())
    return -profit_test[declined].sum(), np.percentile(boots, [2.5, 97.5])

uplift_flat, ci_flat = uplift_ci(policies['Flat profit rule (fit on train)'])
uplift_price, ci_price = uplift_ci(policies['Price-aware profit rule (fit on train)'])
print(f'Flat profit rule vs approve all:  ${uplift_flat / 1e6:+.2f}M (95% CI ${ci_flat[0] / 1e6:+.2f}M to ${ci_flat[1] / 1e6:+.2f}M)')
print(f'Price-aware rule vs approve all: ${uplift_price / 1e6:+.2f}M (95% CI ${ci_price[0] / 1e6:+.2f}M to ${ci_price[1] / 1e6:+.2f}M)')

policy_table.style.format({'approval rate': '{:.1%}', 'bad rate of approved': '{:.1%}',
                           'good loans declined (% of goods)': '{:.1%}', 'bad loans declined (% of bads)': '{:.1%}',
                           'realized profit ($M)': '{:.2f}', 'vs approve all': '{:+.1%}'})

approval_rate = np.arange(1, len(order) + 1) / len(order)
running_bad_rate = np.cumsum(bad_te[order]) / np.arange(1, len(order) + 1)
start = max(1, len(order) // 100)   # skip the first 1%, where the running bad rate is noise

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
axes[0].plot(approval_rate[start:], running_bad_rate[start:], color='grey')
axes[1].plot(approval_rate, cum_profit / 1e6, color='grey')
for name, row in policy_table.iterrows():
    if name.startswith('Best in hindsight'):
        continue
    axes[0].scatter(row['approval rate'], row['bad rate of approved'], zorder=3, label=name)
    axes[1].scatter(row['approval rate'], row['realized profit ($M)'], zorder=3, label=name)
axes[0].set_xlabel('Approval rate')
axes[0].set_ylabel('Bad rate of approved loans')
axes[0].set_title('Approval rate vs bad rate (out-of-time)')
axes[1].set_xlabel('Approval rate')
axes[1].set_ylabel('Realized profit ($M)')
axes[1].set_title('Realized profit by policy (out-of-time)')
axes[1].legend(fontsize=8)
plt.tight_layout();

returns_by_grade = pd.DataFrame({
    'grade': X_test['grade'].astype(str).str.strip(),
    'bad': bad_te,
    'PD': PD_test,
    'profit': profit_test,
    'lent': outcomes.loc[X_test.index, 'funded_amnt'].to_numpy(),
}).groupby('grade').agg(loans=('bad', 'size'), bad_rate=('bad', 'mean'), mean_PD=('PD', 'mean'),
                        profit=('profit', 'sum'), lent=('lent', 'sum'))
returns_by_grade['return_on_lent'] = returns_by_grade['profit'] / returns_by_grade['lent']
returns_by_grade.round(3)

def psi(expected, actual, edges):
    e = np.clip(np.histogram(expected, bins = edges)[0] / len(expected), 1e-6, None)
    a = np.clip(np.histogram(actual, bins = edges)[0] / len(actual), 1e-6, None)
    return float(np.sum((a - e) * np.log(a / e)))

def psi_labels(expected, actual):
    cats = sorted(set(expected) | set(actual), key = str)
    e = np.clip(pd.Series(expected).value_counts(normalize = True).reindex(cats, fill_value = 0).to_numpy(), 1e-6, None)
    a = np.clip(pd.Series(actual).value_counts(normalize = True).reindex(cats, fill_value = 0).to_numpy(), 1e-6, None)
    return float(np.sum((a - e) * np.log(a / e)))

edges = np.unique(np.quantile(scores_train, np.linspace(0, 1, 11)))
edges[0], edges[-1] = -np.inf, np.inf
psi_test, psi_recent = psi(scores_train, scores_test, edges), psi(scores_train, scores_recent, edges)
print(f'Score PSI, train vs {TEST_YEAR}: {psi_test:.3f}')
print(f'Score PSI, train vs {RECENT_YEARS[0]}-{RECENT_YEARS[-1]} applicants: {psi_recent:.3f}')

wb = pipeline['woe']
csi = pd.DataFrame([{
    'feature': f,
    f'CSI vs {TEST_YEAR}': psi_labels(wb.bin_labels(X_train, f), wb.bin_labels(X_test, f)),
    f'CSI vs {RECENT_YEARS[0]}-{RECENT_YEARS[-1]}': psi_labels(wb.bin_labels(X_train, f), wb.bin_labels(X_recent, f)),
} for f in wb.selected_]).set_index('feature').sort_values(f'CSI vs {RECENT_YEARS[0]}-{RECENT_YEARS[-1]}', ascending = False)
display(csi.round(3))

bins = np.linspace(min(scores_train.min(), scores_test.min(), scores_recent.min()),
                   max(scores_train.max(), scores_test.max(), scores_recent.max()), 50)
plt.figure(figsize=(8, 4))
plt.hist(scores_train, bins = bins, density = True, alpha = 0.45, label = f'Train (to {TRAIN_END_YEAR})')
plt.hist(scores_test, bins = bins, density = True, alpha = 0.45, label = f'Out-of-time ({TEST_YEAR})')
plt.hist(scores_recent, bins = bins, density = True, alpha = 0.45, label = f'Applicants {RECENT_YEARS[0]}-{RECENT_YEARS[-1]}')
plt.xlabel('Score')
plt.title('Score distributions')
plt.legend();

score_map = df_scorecard.set_index('Feature name')['Score - Final']
declined = ~policies['Price-aware profit rule (fit on train)']
X_declined = X_test.loc[declined]
dummies = pipeline['woe'].transform(X_declined)

reason_features = [f for f in pipeline['woe'].selected_ if f != 'grade']
shortfall = {}
for f in reason_features:
    cols = [c for c in dummies.columns if c.split(':', 1)[0] == f]
    applicant_points = dummies[cols].to_numpy() @ score_map[cols].to_numpy()   # reference bin scores 0
    best_points = max(0.0, score_map[cols].max())
    shortfall[f] = best_points - applicant_points
shortfall = pd.DataFrame(shortfall, index = X_declined.index)

labels = {f: pipeline['woe'].bin_labels(X_declined, f) for f in reason_features}
top3 = np.argsort(-shortfall.to_numpy(), axis = 1)[:, :3]
reason_rows = []
for r, idx in enumerate(X_declined.index):
    row = {'score': scores_test.loc[idx], 'PD': y_test_proba.loc[idx, 'PD']}
    for k in range(3):
        f = reason_features[top3[r, k]]
        row[f'reason {k + 1}'] = f'{f} {labels[f][r]} (-{shortfall.iloc[r, top3[r, k]]:.0f} pts)'
    reason_rows.append(row)
reason_codes = pd.DataFrame(reason_rows, index = X_declined.index)

print(f'{len(reason_codes):,} out-of-time applicants declined under the price-aware rule')
print('Most common first reason:')
print(reason_codes['reason 1'].str.split(' ').str[0].value_counts(normalize = True).head(6).round(3))
reason_codes.head(10)

print('=' * 70)
print('CREDIT SCORECARD V2: SUMMARY')
print('=' * 70)
print(f'Snapshot {snapshot:%Y-%m}. Population: 36-month loans with final outcomes.')
print(f'Train (to {TRAIN_END_YEAR}): {len(X_train):,} loans, bad rate {1 - y_train.mean():.1%}')
print(f'Out-of-time ({TEST_YEAR}): {len(X_test):,} loans, bad rate {1 - y_test.mean():.1%}')
print(f'Recent applicants ({RECENT_YEARS[0]}-{RECENT_YEARS[-1]}): {len(X_recent):,}')
print()
print('Features:', pipeline['woe'].selected_)
print('Dropped (low IV):', pipeline['woe'].dropped_low_iv_)
print('Dropped (correlated):', pipeline['woe'].dropped_corr_)
print()
print(f'In-time CV Gini {GINI:.3f} | train Gini {gini_train:.3f}')
print(f'Out-of-time Gini {Gini:.3f} (95% CI {gini_ci[0]:.3f} to {gini_ci[1]:.3f}) | KS {KS:.3f} | AUROC {AUROC:.3f}')
print(f'Champion (LendingClub sub-grade) out-of-time Gini {gini_champion:.3f}')
print(f"Calibration: mean predicted PD {calib['PD'].mean():.1%} vs observed {calib['bad'].mean():.1%}")
print('Decile mean PD -> observed: ' + '; '.join(
    f'{r.mean_predicted_PD:.1%}->{r.observed_bad_rate:.1%}' for r in calib_table.itertuples()))
print()
print(f'g {g:.1%}, l {l:.1%} (LGD {lgd:.0%}), flat cutoff PD < {pd_star:.1%}')
print(f'Price-aware cutoff: {price_aware_cutoff(np.array([rates[0]]))[0]:.1%} at {rates[0]:.1f}% rate, '
      f'{price_aware_cutoff(np.array([rates[-1]]))[0]:.1%} at {rates[-1]:.1f}% rate')
print()
print(policy_table.round(4).to_string())
print(f'Flat rule uplift ${uplift_flat / 1e6:+.2f}M (CI {ci_flat[0] / 1e6:+.2f} to {ci_flat[1] / 1e6:+.2f})')
print(f'Price-aware uplift ${uplift_price / 1e6:+.2f}M (CI {ci_price[0] / 1e6:+.2f} to {ci_price[1] / 1e6:+.2f})')
print()
print('Return on lent by grade: ' + '; '.join(
    f'{gr} {r.return_on_lent:+.1%} (bad {r.bad_rate:.0%})' for gr, r in returns_by_grade.iterrows()))
print(f'Score PSI: train vs {TEST_YEAR} {psi_test:.3f}; train vs recent {psi_recent:.3f}')
print('Top first reasons: ' + ', '.join(
    f'{k} {v:.0%}' for k, v in reason_codes['reason 1'].str.split(' ').str[0].value_counts(normalize = True).head(4).items()))
```
